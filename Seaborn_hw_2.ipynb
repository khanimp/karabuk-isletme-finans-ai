```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
sns.set_style('whitegrid')
```

Q2

```python
titanic = sns.load_dataset('titanic')
```

Q3

```python
titanic.head()
```

Q4 plots

```python
sns.jointplot(data=titanic, x='fare', y='age', kind='scatter')
plt.show()
```

```python
sns.histplot(data=titanic, x='fare', bins=30, color='red', alpha=0.4)
plt.show()
```

```python
sns.boxplot(
    data=titanic,
    x='class',
    y='age',
    hue='class',
    palette=['#1f77b4', '#ff7f0e', '#2ca02c'],
    dodge=False,
    legend=False
)
plt.show()
```

```python
sns.swarmplot(
    data=titanic,
    x='class',
    y='age',
    hue='class',
    palette=['#1f77b4', '#ff7f0e', '#2ca02c'],
    size=4,
    legend=False
)
plt.show()
```

```python
sns.countplot(
    data=titanic,
    x='sex',
    hue='sex',
    order=['male', 'female'],
    palette={'male': '#1f77b4', 'female': '#ff7f0e'},
    legend=False
)
plt.show()
```

```python
columns = [
    'survived', 'pclass', 'age', 'sibsp',
    'parch', 'fare', 'adult_male', 'alone'
]

correlation = titanic[columns].corr()

sns.heatmap(correlation, cmap='coolwarm')
plt.show()
```

```python
g = sns.FacetGrid(
    titanic,
    col='sex',
    col_order=['male', 'female']
)

g.map(plt.hist, 'age', bins=10, color='#1f77b4', edgecolor='white')
plt.show()
```
